名前(name)と年齢(age)の属性を持つHumanクラスを作成してください。
nameとageを標準出力(print)するメソッド(printinfo)を追加してください。

Humanクラスのインスタンスは、変数に代入してプログラム内で使用してください。

```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def printinfo(self):
    print(f"名前：{self.name}, 年齢：{self.age}歳")

person1 = Human("侍 太郎", 25)
person2 = Human("侍 花子", 20)

person1.printinfo()  # 出力結果: 名前: 侍 太郎, 年齢: 25歳
person2.printinfo()  # 出力結果: 名前: 侍 花子, 年齢: 20歳

```
